```python
#Examples of try, except, else, and finally in Python

try:
    number = int (input('Enter a number:'))
    result = 100 / number # 

except ValueError:
    print ("Error: Cannot divide by 0") # code that runs if error occurrs

# Example 2 :using this when knowing user may input values that result in errors
except ZeroDivisionError:
    print ('Cannot divide by zero')

#example 3 - Handling errors using try/except with loops
while True:
    try:
        number = int(input('Enter a number '))
        if number > 0:
            break
        else:
            print("Number must be a positive.")
    except ValueError:
        print('Invalid input. Please enter an integer')

#handling error 

'''
try:
    code that may/may not produce errors
except:
    when error arises, then this block of code executes 
else:
    when error doesn't arise, then this block of code executes
finally:
    this block will execute whether error occurs or not.
''' 

try:
    a = int (input())
    b = int (input())
    result = a/b 
    print (result)

except:
    print ('We caught an error')
else:
    print ('Hurray, we dont have any errors!')
finally:
    print ('I have reached the end of the line.')
```
