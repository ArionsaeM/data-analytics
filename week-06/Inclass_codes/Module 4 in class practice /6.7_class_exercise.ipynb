```python
#Arionsae McCutcheon
# Product Inventory Tracker
print ()
class Product:
    def __init__(self ,name = "Unknown", category = "Unknown", price = 0 , quantity = 0):
        
        self.name = name
        self.category = category
        self.price = price
        self.quantity = quantity

    def display_info (self):
            print (f'Product Name: {self.name}')
            print(f'Product Category: {self.category}')
            print (f'Product Price: {self.price}')
            print (f'Product Quantity: {self.quantity}')
            print ('{'-' * 20}')
    def inventory_value (self):
        self.inventory_value = price*quantity
        print (f'Inventory Value: {self.inventory_value}')

product1 = Product ('Chair', 'Furniture', '50.00', '4')
product2 = Product ('Laptop', 'Tech', '200.00', '2')

product1.display_info
product2.display_info
```
