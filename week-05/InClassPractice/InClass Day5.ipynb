```python
day = 1 

monday = 1200
tuesday  = 1500
wednesday = 1100
thursday = 1800
friday = 1400

total_sales = 0

for day in range(1, 6): #i already changed from a while loop to a for loop
    if monday > 0:
        total_sales = monday + total_sales
    elif tuesday > 0:
        total_sales = tuesday + total_sales
        day += 1
    elif wednesday < 0:
        total_sales = wednesday + total_sales
        day += 1
    elif thursday < 0:
        total_sales = thursday + total_sales
        day += 1
    else:
        total_sales = friday + total_sales
        day += 1
print (f'Total Sales for the week: {total_sales}')
print(f'Average Sales for the week: {total_sales / 5}')

```
